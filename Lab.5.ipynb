1. Create a `matrix` class with the following properties:
    * It can be initialized in 2 ways:
        1. with arguments `n` and `m`, the size of the matrix. A newly instanciated matrix will contain all zeros.
        2. with a list of lists of values. Note that since we are using lists of lists to implement matrices, it is possible that not all rows have the same number of columns. Test explicitly that the matrix is properly specified.
    * Matrix instances `M` can be indexed with `M[i][j]` and `M[i,j]`.
    * Matrix assignment works in 2 ways:
        1. If `M_1` and `M_2` are `matrix` instances `M_1=M_2` sets the values of `M_1` to those of `M_2`, if they are the same size. Error otherwise.
        2. In example above `M_2` can be a list of lists of correct size.


```python
class Matrix:
    def __init__(self, n, m=None):
        if isinstance(n, int) and isinstance(m, int):
            self.n = n
            self.m = m
            self.matrix = [[0] * self.m for _ in range(self.n)]
        elif isinstance(n, list) and m is None:
            row_length = len(n[0]) if n else 0
            for row in n:
                if len(row) != row_length:
                    raise ValueError("Rows must have same number of columns")
            self.n = len(n)
            self.m = row_length
            self.matrix = [row[:] for row in n]
        else:
            raise ValueError("Initialize with (n, m) or list of lists only")

    def __getitem__(self, index):
        if isinstance(index, tuple):
            i, j = index
            return self.matrix[i][j]
        return self.matrix[index]

    def __setitem__(self, index, value):
        if isinstance(index, tuple):
            i, j = index
            self.matrix[i][j] = value
        else:
            if len(value) != self.m:
                raise ValueError("Row must have sanme number of columns")
            self.matrix[index] = value

    def assign(self, other):
        if isinstance(other, Matrix):
            if self.n == other.n and self.m == other.m:
                self.matrix = [row[:] for row in other.matrix]
            else:
                raise ValueError("Matrix dimensions do not match")
        elif isinstance(other, list):
            if len(other) != self.n or any(len(row) != self.m for row in other):
                raise ValueError("Matrix dimensions do not match")
            self.matrix = [row[:] for row in other]
        else:
            raise TypeError("Assignment must be a (nxm) matrix or a list of lists")

    def __repr__(self):
        return '\n'.join(str(row) for row in self.matrix)
```

```python
class Matrix:
    def __init__(self, n, m=None):
        if isinstance(n, int) and isinstance(m, int):
            self.n = n
            self.m = m
            self.matrix = [[0] * self.m for _ in range(self.n)]
        elif isinstance(n, list) and m is None:
            row_length = len(n[0]) if n else 0
            for row in n:
                if len(row) != row_length:
                    raise ValueError("Rows must have same number of columns")
            self.n = len(n)
            self.m = row_length
            self.matrix = [row[:] for row in n]
        else:
            raise ValueError("Initialize with (n, m) or list of lists only")

    def __getitem__(self, index):
        if isinstance(index, tuple):
            i, j = index
            if isinstance(i, slice) or isinstance(j, slice):
                result = [
                    row[j] if isinstance(j, slice) else [row[j]]
                    for row in self.matrix[i]
                ]
                return Matrix(result)
            else:
                return self.matrix[i][j]
        elif isinstance(index, int):
            return self.matrix[index]
        else:
            raise TypeError("Invalid index type")

    def __setitem__(self, index, value):
        if isinstance(index, tuple):
            i, j = index
            self.matrix[i][j] = value
        else:
            if len(value) != self.m:
                raise ValueError("Row must have sanme number of columns")
            self.matrix[index] = value

    def assign(self, other):
        if isinstance(other, Matrix):
            if self.n == other.n and self.m == other.m:
                self.matrix = [row[:] for row in other.matrix]
            else:
                raise ValueError("Matrix dimensions do not match")
        elif isinstance(other, list):
            if len(other) != self.n or any(len(row) != self.m for row in other):
                raise ValueError("Matrix dimensions do not match")
            self.matrix = [row[:] for row in other]
        else:
            raise TypeError("Assignment must be a (nxm) matrix or a list of lists")

    def __repr__(self):
        return '\n'.join(str(row) for row in self.matrix)

    def shape(self):
        return (self.n, self.m)

    def transpose(self):
        transposed = [[self.matrix[i][j] for i in range(self.n)] for j in range(self.m)]
        return Matrix(transposed)

    def row(self,r):
        if r < 0 or r >= self.n:
            raise IndexError("Row index out of range")
        return Matrix([self.matrix[r]])

    def row(self,r):
        if r < 0 or r >= self.n:
            raise IndexError("Row index out of range")
        return Matrix([self.matrix[r]])

    def column(self, c):
        if c < 0 or c >= self.m:
            raise IndexError("Column index out of range")
        return Matrix([[self.matrix[i][c]] for i in range(self.n)])

    def to_list(self):
        return [row[:] for row in self.matrix]

    def block(self, n_0, n_1, m_0, m_1):
        if not (0 <= n_0 <= n_1 <= self.n) or not (0 <= m_0 <= m_1 <= self.m):
            raise IndexError("Block indices out of range")
        submatrix = [row[m_0:m_1] for row in self.matrix[n_0:n_1]]
        return Matrix(submatrix)
```

3. Write functions that create special matrices (note these are standalone functions, not member functions of your `matrix` class):
    * `constant(n,m,c)`: returns a `n` by `m` matrix filled with floats of value `c`.
    * `zeros(n,m)` and `ones(n,m)`: return `n` by `m` matrices filled with floats of value `0` and `1`, respectively.
    * `eye(n)`: returns the n by n identity matrix.

```python
def constant(n,m,c):
    return Matrix([[float(c)] * m for _ in range(n)])
```

```python
def zeros(n,m):
    return Matrix([[0.0] * m for _ in range(n)])
```

```python
def ones(n,m):
    return Matrix([[1.0] * m for _ in range(n)])
```

```python
def eye(n):
    return Matrix([[1.0 if i == j else 0.0 for j in range(n)] for i in range(n)])
```

```python
class Matrix:
    def __init__(self, n, m=None):
        if isinstance(n, int) and isinstance(m, int):
            self.n = n
            self.m = m
            self.matrix = [[0] * self.m for _ in range(self.n)]
        elif isinstance(n, list) and m is None:
            row_length = len(n[0]) if n else 0
            for row in n:
                if len(row) != row_length:
                    raise ValueError("Rows must have same number of columns")
            self.n = len(n)
            self.m = row_length
            self.matrix = [row[:] for row in n]
        else:
            raise ValueError("Initialize with (n, m) or list of lists only")

    def __getitem__(self, index):
        if isinstance(index, tuple):
            i, j = index
            if isinstance(i, slice) or isinstance(j, slice):
                result = [
                    row[j] if isinstance(j, slice) else [row[j]]
                    for row in self.matrix[i]
                ]
                return Matrix(result)
            else:
                return self.matrix[i][j]
        elif isinstance(index, int):
            return self.matrix[index]
        else:
            raise TypeError("Invalid index type")

    def __setitem__(self, index, value):
        if isinstance(index, tuple):
            i, j = index
            self.matrix[i][j] = value
        else:
            if len(value) != self.m:
                raise ValueError("Row must have sanme number of columns")
            self.matrix[index] = value

    def assign(self, other):
        if isinstance(other, Matrix):
            if self.n == other.n and self.m == other.m:
                self.matrix = [row[:] for row in other.matrix]
            else:
                raise ValueError("Matrix dimensions do not match")
        elif isinstance(other, list):
            if len(other) != self.n or any(len(row) != self.m for row in other):
                raise ValueError("Matrix dimensions do not match")
            self.matrix = [row[:] for row in other]
        else:
            raise TypeError("Assignment must be a (nxm) matrix or a list of lists")

    def __repr__(self):
        return '\n'.join(str(row) for row in self.matrix)

    def shape(self):
        return (self.n, self.m)

    def transpose(self):
        transposed = [[self.matrix[i][j] for i in range(self.n)] for j in range(self.m)]
        return Matrix(transposed)

    def row(self,r):
        if r < 0 or r >= self.n:
            raise IndexError("Row index out of range")
        return Matrix([self.matrix[r]])

    def row(self,r):
        if r < 0 or r >= self.n:
            raise IndexError("Row index out of range")
        return Matrix([self.matrix[r]])

    def column(self, c):
        if c < 0 or c >= self.m:
            raise IndexError("Column index out of range")
        return Matrix([[self.matrix[i][c]] for i in range(self.n)])

    def to_list(self):
        return [row[:] for row in self.matrix]

    def block(self, n_0, n_1, m_0, m_1):
        if not (0 <= n_0 <= n_1 <= self.n) or not (0 <= m_0 <= m_1 <= self.m):
            raise IndexError("Block indices out of range")
        submatrix = [row[m_0:m_1] for row in self.matrix[n_0:n_1]]
        return Matrix(submatrix)

    def scalarmul(self, c):
        return Matrix([[self.matrix[i][j] * c for j in range(self.m)] for i in range(self.n)])

    def add(self, other):
        if self.shape() != other.shape():
            raise ValueError("Matrix sizes do not match")
        return Matrix([[self.matrix[i][j] + other[i, j] for j in range(self.m)] for i in range(self.n)])

    def sub(self, other):
        if self.shape() != other.shape():
            raise ValueError("Matrix sizes do not match")
        return Matrix([[self.matrix[i][j] - other[i, j] for j in range(self.m)] for i in range(self.n)])

    def mat_mult(self, other):
        if self.m != other.n:
            raise ValueError("Matrix sizes do not match")
        return Matrix([
            [sum(self.matrix[i][k] * other[k, j] for k in range(self.m)) for j in range(other.m)]
            for i in range(self.n)
        ])

    def element_mult(self, other):
        if self.shape() != other.shape():
            raise ValueError("Matrix sizes do not match")
        return Matrix([[self.matrix[i][j] * other[i, j] for j in range(self.m)] for i in range(self.n)])

    def equals(self, other):
        if self.shape() != other.shape():
            return False
        return all(self.matrix[i][j] == other[i, j] for i in range(self.n) for j in range(self.m))
```
